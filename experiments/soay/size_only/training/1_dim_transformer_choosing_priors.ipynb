```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D
from matplotlib.patches import Patch

plt.style.use("seaborn-v0_8-white")

years = np.sort(soay_data["yr"].dropna().unique())
year_colors = sns.color_palette("viridis", n_colors=len(years))
pooled_color = "#D55E00"

fig, axis = plt.subplots(figsize=(8, 4.5), constrained_layout=True)

sns.kdeplot(x=pooled_sizes, fill=True, cut=0, color=pooled_color, linewidth=0, alpha=0.15, ax=axis)

sns.kdeplot(data=soay_data, x="z", hue="yr", hue_order=years, common_norm=False, cut=0, palette=year_colors, linewidth=1.2, alpha=0.7, legend=False, ax=axis)

sns.kdeplot(x=pooled_sizes, cut=0, color=pooled_color, linewidth=2.2, ax=axis)

normal_mean = 3
normal_variance = 0.5**2
normal_x = np.linspace(*axis.get_xlim(), 500)
normal_density = np.exp(-(normal_x - normal_mean) ** 2 / (2 * normal_variance)) / np.sqrt(2 * np.pi * normal_variance)

axis.plot(normal_x, normal_density, color="#333333", linestyle="--", linewidth=2)

reference_handles = [
    Patch(facecolor=pooled_color, edgecolor=pooled_color, alpha=0.3, label="Pooled Density"),
    Line2D([0], [0], color="#333333", linestyle="--", linewidth=2, label=r"Normal ($\mu=3,\ \sigma^2=0.5^2$) Prior")
]

reference_legend = axis.legend(handles=reference_handles, loc="upper right", frameon=False)
axis.add_artist(reference_legend)

year_handles = [Line2D([0], [0], color=color, linewidth=1.5, label=str(int(year))) for year, color in zip(years, year_colors)]

axis.legend(handles=year_handles, title="Year", bbox_to_anchor=(1.02, 0.5), loc="center left", ncol=2 if len(years) > 8 else 1, frameon=False)

axis.set_xlabel("Size", fontsize=14)
axis.set_ylabel("Density", fontsize=14)
axis.set_title("Pooled And Annual Soay Sheep Size Densities", fontsize=16)
axis.tick_params(labelsize=9, direction="out")
axis.grid(False)
axis.spines["top"].set_visible(False)
axis.spines["right"].set_visible(False)

plt.savefig("soay_sheep_density_overlay.pdf")
plt.savefig("soay_sheep_density_overlay.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
with np.load(GENERATED / 'soay/size_only/inference/gpdf_gp_samples.npz') as saved:
    gp_samples = {name: saved[name] for name in saved.files}
```

```python
labels = {"surv_ls": "Survival", 
          "repr_ls": "Reproduction", 
          "growth_ls": "Growth", 
          "rcsz_ls": "Recruit Size"}

lengthscale_samples = {labels[name]: values for name, values in gp_samples.items() if name.endswith("_ls")}
figure, axes = plt.subplots(2, 2, figsize=(11, 7))

for axis, (name, values) in zip(axes.ravel(), lengthscale_samples.items()):
    sns.kdeplot(values, ax=axis, fill=True, cut=0)
    axis.set_xlabel("Lengthscale")
    axis.set_ylabel("Density")
    axis.set_title(f"{name} Lengthscale")
    axis.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import lognorm

plt.style.use("seaborn-v0_8-white")

pooled_lengthscales = np.concatenate(list(lengthscale_samples.values()))

target_mean = 2.0
target_variance = 4.0
log_variance = np.log(1 + target_variance / target_mean**2)
log_sd = np.sqrt(log_variance)
log_mean = np.log(target_mean) - log_variance / 2

x = np.linspace(0.0001, 10, 1000)
lognormal_density = lognorm.pdf(x, s=log_sd, scale=np.exp(log_mean))

posterior_color = "#D55E00"
prior_color = "#333333"

fig, axis = plt.subplots(figsize=(8, 4.5), constrained_layout=True)

sns.kdeplot(x=pooled_lengthscales, fill=True, cut=0, color=posterior_color, linewidth=0, alpha=0.2, ax=axis)
sns.kdeplot(x=pooled_lengthscales, cut=0, color=posterior_color, linewidth=2, label="Pooled Posterior", ax=axis)

axis.plot(x, lognormal_density, color=prior_color, linestyle="--", linewidth=2, label=r"Log-Normal Prior ($\mu=2,\ \sigma^2=4$)")

axis.set_xlim(0, 10)
axis.set_xlabel("Lengthscale", fontsize=14)
axis.set_ylabel("Density", fontsize=14)
axis.set_title("Pooled Lengthscale Posterior And Log-Normal Prior", fontsize=16)
axis.tick_params(labelsize=9, direction="out")
axis.grid(False)
axis.spines["top"].set_visible(False)
axis.spines["right"].set_visible(False)
axis.legend(frameon=False, fontsize=10)

plt.savefig("lengthscale_prior_posterior.pdf")
plt.savefig("lengthscale_prior_posterior.png", dpi=300, bbox_inches="tight")
plt.show()
```
