```python
import numpy as np
import jax.numpy as jnp
import matplotlib.pyplot as plt
from jax import random
from dl4bi_sps.kernels import rbf

def draw_gp_prior(seed=1, 
                  num_samples=5, 
                  lengthscale=1.0, 
                  variance=1.0, 
                  num_points=100, 
                  jitter=5e-4):
    locations = jnp.linspace(1.0, 5.0, num_points).reshape(-1, 1)
    covariance = rbf(locations, locations, variance, lengthscale) + jitter * jnp.eye(num_points)
    z = random.normal(random.key(seed), (num_samples, num_points))
    samples = jnp.einsum("ij,bj->bi", jnp.linalg.cholesky(covariance), z)
    locations = np.asarray(locations).ravel()
    samples = np.asarray(samples)

    return locations, samples
```

```python
plt.style.use("seaborn-v0_8-white")
plt.rcParams.update({"font.family": "Arial", "mathtext.fontset": "cm"})

lengthscales = [0.1, 1.0, 5.0]
variances = [1.0, 5.0]
colors = ["#0072B2", "#D55E00", "#009E73"]

fig, axes = plt.subplots(2, 3, figsize=(5.906, 5.0), sharex=True, sharey=True, constrained_layout=True)

for row, variance in enumerate(variances):
    for column, (lengthscale, color) in enumerate(zip(lengthscales, colors)):
        axis = axes[row, column]
        locations, samples = draw_gp_prior(seed=1, lengthscale=lengthscale, variance=variance)

        for sample in samples:
            axis.plot(locations, sample, color=color, linewidth=1.0, alpha=0.4)

        if row == 0:
            axis.set_title(rf"$\ell = {lengthscale}$", fontsize=11)

        axis.grid(False)
        axis.tick_params(labelsize=9, direction="out")
        axis.spines["top"].set_visible(False)
        axis.spines["right"].set_visible(False)

    axes[row, -1].annotate(rf"$\sigma_f^2 = {variance}$", xy=(1.07, 0.5), xycoords="axes fraction", va="center", fontsize=10, annotation_clip=False)

fig.supxlabel("Input Location", fontsize=11)
fig.supylabel("Function Value", fontsize=11)

fig.savefig("gp_prior_lengthscales_variances.pdf", bbox_inches="tight")
fig.savefig("gp_prior_lengthscales_variances.png", dpi=300, bbox_inches="tight")
plt.show()
```
