# In this codebook given posterior samples, we make a series of IPMs and plot results.

```python
import numpy as np

from ipm_utils import (
    mk_K_glm, 
    growth_fn_glm, 
    flowering_fn_glm, 
    survival_fn_glm, 
    seed_fn_glm, 
    new_recruits_fn_glm
    )
```

```python
with np.load("gpdf_glm_samples.npz") as data:
    glm_samples = {name: data[name] for name in data.files}

with np.load("gpdf_gp_samples.npz") as data:
    gp_samples = {name: data[name] for name in data.files}

L = -2.5
U = 4
N_mesh = 250 
N_samples = 10
```

```python
glm_samples.keys()
```

```python
p_recr_est = float(glm_samples.pop('recr_prob'))
recr_mean_est = float(glm_samples.pop('recr_mean'))
recr_sd_est = float(glm_samples.pop('recr_sd'))

fixed_param = {
'p.r': p_recr_est,
'rcsz.int':recr_mean_est,
'rcsz.sd': recr_sd_est,
}
```

```python
glm_funcs = {
    'repr': flowering_fn_glm, 
    'surv': survival_fn_glm,
    'seed': seed_fn_glm,
    'grow': growth_fn_glm,
    'new_recruits_fn': new_recruits_fn_glm
    }
```

```python
def extract_glm_samples(glm_samples, index):
    draw = {}
    for key, item in glm_samples.items():
        if item.ndim != 0:
            draw[key] = item[index]
    return draw
```

```python
glm_ipm_results =[]
for i in range(N_samples + 1):
    temp_dict = extract_glm_samples(glm_samples, i)
    par_dict = fixed_param | temp_dict
    result = mk_K_glm(
        n_mesh_points = N_mesh, 
        m_par = par_dict,
        lower_size = L,
        upper_size = U,
        vital_rate_functions = glm_funcs, 
        correction = True
        )
    glm_ipm_results.append(result)
```
