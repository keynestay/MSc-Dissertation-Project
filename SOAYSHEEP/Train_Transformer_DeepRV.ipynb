# Train Transformer DeepRV For Soay Sheep

This notebook uses DeepRV's original transformer model, training loop, masked loss, validation, optimizer, and checkpoint utilities. The dataloader is changed only to sample one-dimensional Soay locations from `[1, 5]`, lengthscales from `[0.05, 10]`, and active sequence lengths uniformly from `64` to `2048`.

```python
import sys
from pathlib import Path

repo_root = Path.cwd().parent if Path.cwd().name == "SOAYSHEEP" else Path.cwd()
sys.path.insert(0, str(repo_root))

from SOAYSHEEP.train_transformer_deeprv import TrainConfig, load_checkpoint, train_model
```

```python
config = TrainConfig(output_dir=str(repo_root / "SOAYSHEEP" / "results" / "transformer_deeprv"), min_locations=64, max_locations=2048, size_min=1.0, size_max=5.0, lengthscale_min=0.05, lengthscale_max=10.0, batch_size=8, num_steps=2_000_000, checkpoint_interval=100_000, validation_interval=400_000, validation_steps=10_000, learning_rate=1e-4, clip_norm=3.0, wandb_mode="disabled")
config
```

`latest.ckpt` is overwritten every 100,000 steps. Following the original DeepRV convention, `model.ckpt` stores the best validation state; `final.ckpt` stores the state from the last training step.

```python
result = train_model(config)
result["metadata"]
```

Load either the best-validation checkpoint (`model.ckpt`) or the final training state (`final.ckpt`) with the helper below.

```python
model, final_state = load_checkpoint(Path(config.output_dir) / "final.ckpt", config)
int(final_state.step)
```
